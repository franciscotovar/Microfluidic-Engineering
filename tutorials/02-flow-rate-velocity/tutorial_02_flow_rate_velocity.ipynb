# Tutorial 2: Flow Rate and Velocity
#### Does the same flow rate mean the same flow?

##### Microfluidic engineering | Dr. Francisco Tovar

**Goal:** Understand the difference between volumetric flow rate and fluid velocity, and explore how channel geometry transforms an imposed syringe-pump flow rate into the velocity experienced inside a microfluidic device. **No CFD in this tutorial.**

## 1. Start with the engineering problem

In many microfluidic experiments, we control the system using a syringe pump.

The pump display may say:

$$
Q=100~\mu\mathrm{L/min}.
$$

It is tempting to think that this tells us how fast the fluid is moving. It does not.

The syringe pump controls **volumetric flow rate**, $Q$: the volume delivered per unit time.

Velocity, $U$, describes how fast fluid moves through a particular cross-section.

$$
Q=UA
$$

and therefore

$$
U=\frac{Q}{A}.
$$

This gives us the engineering question:

> **My syringe pump says 100 µL/min. What is the velocity inside my microchannel?**

We cannot answer until we know the channel cross-sectional area.

## 2. Flow rate is not velocity

Flow rate and velocity are related, but they are different physical quantities.

### Volumetric flow rate

$$
Q=\frac{\mathrm{volume}}{\mathrm{time}}
$$

Typical units: $\mu$L/min or mL/min.

### Mean velocity

$$
U=\frac{\mathrm{distance}}{\mathrm{time}}
$$

Typical units: m/s or mm/s.

For a given flow rate,

$$
U=\frac{Q}{A}.
$$

Therefore, the same syringe-pump setting can produce very different velocities in channels with different cross-sectional areas.

> **The pump controls $Q$. The geometry helps determine $U$.**

## 3. Why should an engineer care?

One word: **Reproducibility.**

Two studies may both report

$$
Q=100~\mu\mathrm{L/min},
$$

but if their channel dimensions are different, their mean velocities can be very different.

Velocity also enters later calculations of Reynolds number, residence time, shear rate, shear stress, and transport.

So flow rate and channel geometry belong together when describing a microfluidic experiment.

## 4. Start with a rectangular microchannel

For a rectangular channel,

$$
A=wh.
$$

Use the same example geometry as Tutorial 1:

$$
w=200~\mu\mathrm{m},
\qquad
h=85~\mu\mathrm{m}.
$$

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

w_um = 200
h_um = 85

w = w_um * 1e-6
h = h_um * 1e-6
A = w * h

print(f"Width  = {w_um} µm")
print(f"Height = {h_um} µm")
print(f"Area   = {A:.3e} m²")
```

## 5. Worked example: 100 µL/min

Suppose

$$
Q=100~\mu\mathrm{L/min}.
$$

Convert to SI units using

$$
1~\mu\mathrm{L}=10^{-9}~\mathrm{m^3}
$$

and

$$
1~\mathrm{min}=60~\mathrm{s}.
$$

Then calculate

$$
U=\frac{Q}{A}.
$$

```python
Q_uL_min = 100
Q = Q_uL_min * 1e-9 / 60
U = Q / A

print(f"Flow rate     = {Q_uL_min} µL/min")
print(f"Flow rate     = {Q:.3e} m³/s")
print(f"Mean velocity = {U:.4f} m/s")
print(f"Mean velocity = {U*1000:.1f} mm/s")
```

The mean velocity is approximately

$$
U\approx98~\mathrm{mm/s}.
$$

The important point is not the number itself.

**The velocity could not be determined from the pump setting alone.**

## 6. Same flow rate, different channel sizes

Keep

$$
Q=100~\mu\mathrm{L/min}
$$

constant and change only the channel cross-section.

Before calculating anything:

> **If the channel becomes smaller while $Q$ remains constant, what should happen to the mean velocity?**

```python
channels = pd.DataFrame({
    "Width [µm]":  [1000, 500, 300, 200, 100],
    "Height [µm]": [500, 250, 150, 85, 50]
})

channels["Area [µm²]"] = channels["Width [µm]"] * channels["Height [µm]"]
channels["Area [m²]"] = channels["Area [µm²]"] * 1e-12
channels["U [m/s]"] = Q / channels["Area [m²]"]
channels["U [mm/s]"] = channels["U [m/s]"] * 1000

display(channels[["Width [µm]","Height [µm]","Area [µm²]","U [mm/s]"]].round(2))
```

```python
fig, ax = plt.subplots(figsize=(8,5))
ax.plot(channels["Area [µm²]"], channels["U [mm/s]"], marker="o")
ax.set(
    xlabel="Channel cross-sectional area [µm²]",
    ylabel="Mean velocity [mm/s]",
    title="Same flow rate, different channel areas"
)
ax.grid(alpha=.3)
plt.tight_layout()
plt.show()
```

Because

$$
U=\frac{Q}{A},
$$

reducing the area increases the mean velocity.

This is why reporting only “100 µL/min” does not completely specify the local flow condition.

## 7. What happens when the area changes inside one device?

The same idea applies within a single device:

**wide channel → contraction → narrow channel → expansion**

For incompressible flow,

$$
Q_1=Q_2=Q_3,
$$

but the local mean velocity is

$$
U=\frac{Q}{A}.
$$

The pump setting has not changed.

**The geometry has changed the local velocity.**

## 8. Example: a contraction inside the device

At the same flow rate, compare:

| Section | Width | Height |
|---|---:|---:|
| Upstream | 400 µm | 100 µm |
| Contraction | 100 µm | 50 µm |
| Downstream | 200 µm | 85 µm |

```python
sections = pd.DataFrame({
    "Section": ["Upstream", "Contraction", "Downstream"],
    "Width [µm]": [400, 100, 200],
    "Height [µm]": [100, 50, 85]
})

sections["Area [µm²]"] = sections["Width [µm]"] * sections["Height [µm]"]
sections["Area [m²]"] = sections["Area [µm²]"] * 1e-12
sections["U [mm/s]"] = (Q / sections["Area [m²]"]) * 1000

display(sections[["Section","Width [µm]","Height [µm]","Area [µm²]","U [mm/s]"]].round(1))
```

```python
fig, ax = plt.subplots(figsize=(8,5))
ax.bar(sections["Section"], sections["U [mm/s]"])
ax.set(
    xlabel="Device section",
    ylabel="Mean velocity [mm/s]",
    title="Velocity changes inside the same device"
)
ax.grid(axis="y", alpha=.3)
plt.tight_layout()
plt.show()
```

## 9. Mean velocity does not mean uniform velocity

When we calculate

$$
U=\frac{Q}{A},
$$

we obtain the **cross-sectional mean velocity**.

This does not mean that every point in the channel has velocity $U$.

Because of no-slip, velocity is zero at stationary walls and higher away from them. A developed laminar flow therefore has a non-uniform velocity profile.

$U=Q/A$ is an average description, not the complete velocity field.

## 10. Connect geometry to Reynolds number

Tutorial 1 introduced

$$
Re=\frac{\rho U D_h}{\mu}.
$$

For a rectangular channel,

$$
D_h=\frac{2wh}{w+h}.
$$

The pump sets $Q$, but geometry changes both $A$ and $D_h$. Therefore, even at the same pump setting, changing channel dimensions can change Reynolds number.

```python
rho = 1000
mu = 1e-3

channels["Dh [m]"] = (
    2 * (channels["Width [µm]"]*1e-6) * (channels["Height [µm]"]*1e-6)
    / ((channels["Width [µm]"]*1e-6) + (channels["Height [µm]"]*1e-6))
)

channels["Re"] = rho * channels["U [m/s]"] * channels["Dh [m]"] / mu

display(channels[["Width [µm]","Height [µm]","U [mm/s]","Re"]].round(2))
```

## 11. Now vary the syringe-pump setting

Keep the geometry fixed at

$$
200~\mu\mathrm{m}\times85~\mu\mathrm{m}
$$

and vary $Q$.

Since $A$ is constant,

$$
U=\frac{Q}{A},
$$

so mean velocity should increase linearly with flow rate.

```python
flow_rates = np.array([10, 25, 50, 100, 200, 400])

Q_values = flow_rates * 1e-9 / 60
U_values = Q_values / A

flow_table = pd.DataFrame({
    "Q [µL/min]": flow_rates,
    "U [mm/s]": U_values * 1000
})

display(flow_table.round(2))

fig, ax = plt.subplots(figsize=(8,5))
ax.plot(flow_table["Q [µL/min]"], flow_table["U [mm/s]"], marker="o")
ax.set(
    xlabel="Flow rate [µL/min]",
    ylabel="Mean velocity [mm/s]",
    title="Flow rate and mean velocity in a fixed channel"
)
ax.grid(alpha=.3)
plt.tight_layout()
plt.show()
```

## 12. What should we report?

If velocity matters to the experiment, reporting only the syringe-pump setting may not be sufficient.

At minimum, provide enough information to reconstruct the mean velocity:

- flow rate, $Q$;
- channel width, $w$;
- channel height, $h$.

Then

$$
A=wh
$$

and

$$
U=\frac{Q}{A}.
$$

For Reynolds number, density and viscosity are also required.

## 14. Take-home message

A syringe pump controls **flow rate**, not velocity.

$$
\boxed{U=\frac{Q}{A}}
$$

Therefore:

> **The same flow rate does not necessarily mean the same flow condition.**

A change in channel dimensions can change velocity even when the pump setting remains unchanged.
